As part of the internship test, I worked on a small data processing task
where I merged multiple datasets related to food delivery.
The goal was to combine orders, users, and restaurant data into a final dataset.

```python
# firstly Import required libraries
import pandas as pd
import sqlite3
# Read input datasets
orders = pd.read_csv("orders.csv")
users = pd.read_json("users.json")
# Connect to SQLite database
conn = sqlite3.connect("restaurants.db")
# Load SQL script to create restaurants table
with open("restaurants.sql", "r") as f:
    sql_script = f.read()
conn.executescript(sql_script)
# Read restaurants data from database
restaurants = pd.read_sql("SELECT * FROM restaurants", conn)
# Merge datasets step by step
merged = orders.merge(users, on="user_id", how="left")
final_df = merged.merge(restaurants, on="restaurant_id", how="left")
# Save final dataset
final_df.to_csv("final_food_delivery_dataset.csv", index=False)
print("Final dataset created successfully")

```
